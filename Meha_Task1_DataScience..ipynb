```python
# 1. INTEGERS, FLOATS, STRINGS, BOOLEAN - Real world data science example
age = 22
salary = 45000.50
name = "Meha"
is_student = True

print("--- Basic Data Types ---")
print(f"Age: {age}, Type: {type(age)}")
print(f"Salary: {salary}, Type: {type(salary)}")
print(f"Name: {name}, Type: {type(name)}")
print(f"Is Student: {is_student}, Type: {type(is_student)}")
```

```python
# 2. LIST, TUPLE, SET, DICTIONARY - Data Science examples
# List - mutable, scores can change
scores = [85, 90, 78, 92, 88]
print("Scores (List):", scores, type(scores))

# Tuple - immutable, fixed info like DOB
dob = (2003, 5, 15)
print("DOB (Tuple):", dob, type(dob))

# Set - unique values, e.g., unique skills
skills = {"Python", "Pandas", "SQL", "Python"}
print("Skills (Set - unique only):", skills, type(skills))

# Dictionary - key-value, student record
student = {
    "name": "Meha",
    "age": 22,
    "department": "Data Science",
    "cgpa": 8.5
}
print("Student Record (Dict):", student, type(student))
print("CGPA:", student["cgpa"])
```

```python
# 3. TYPE CONVERSION & BASIC OPERATIONS
print("--- Type Conversion ---")
# String to number - very important in data science
age_str = "22"
age_int = int(age_str)
print(f"String '{age_str}' converted to int: {age_int}, Type: {type(age_int)}")

salary_str = "45000.50"
salary_float = float(salary_str)
print(f"String '{salary_str}' to float: {salary_float}")

# Number to string
cgpa = 8.5
cgpa_str = str(cgpa)
print(f"Float {cgpa} to string: '{cgpa_str}'")

print("\n--- Basic Operations ---")
# Operations with real-world data
total_score = sum(scores)
avg_score = total_score / len(scores)
print(f"Total Score: {total_score}, Average: {avg_score:.2f}")

# String operations
full_name = "Meha" + " " + "Data Science Intern"
print("Full Title:", full_name)
print("Name in uppercase:", name.upper())

# Boolean logic
is_eligible = (age >= 21) and (student["cgpa"] > 8.0)
print(f"Eligible for Internship? {is_eligible}")

# List operations
scores.append(95)
print("After adding new score:", scores)
print("Highest score:", max(scores))
```

```python
# 4. COMPLETE REAL-WORLD EXAMPLE
print("=== STUDENT PERFORMANCE SUMMARY ===")
students_data = [
    {"name": "Meha", "age": 22, "scores": [85, 90, 88]},
    {"name": "Arun", "age": 23, "scores": [78, 82, 80]}
]

for s in students_data:
    avg = sum(s["scores"]) / len(s["scores"])
    print(f"{s['name']} (Age: {s['age']}) - Avg Score: {avg:.1f} - Type of avg: {type(avg)}")

print("\nTask 1 Completed Successfully!")
```
